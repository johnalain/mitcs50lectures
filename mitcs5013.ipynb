```python
def divide_nums():
    try:
        a = int(input("tell me one number: "))
        b = int(input("tell me another number: "))
        print("a+b = ",a/b)
        print("a/b = ",a/b)
    except ValueError:
        print("COULDN'T convert to number")
    except ZeroDivisionError:
        print("can't divide by zero")
        print("a/b = infinity")
        print("a+b = ",a + b) # type: ignore
    except:
        print("something went very wrong")
    else:
        print("success")
divide_nums()
```

```python
class_list = [
    [['rita', 'michel'], [80.0, 70.0, 85.0]],
    [['josephine', 'kadi'], [100.0, 80.0, 74.0]]
]

def avg(grades):
    return sum(grades) / len(grades)

def get_stats(class_list):
    new_stats = []
    for stu in class_list:
        # stu[0] = names, stu[1] = grades
        new_stats.append([stu[0], stu[1], avg(stu[1])])
    return new_stats

print(get_stats(class_list))
```
