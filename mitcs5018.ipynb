
Typical implementations create a new instance of the class by invoking the superclass’s __new__() method using super().__new__(cls[, ...]) examples please

Great question — this is about how object creation works at a lower level in Python.

__new__ is responsible for creating the instance, while __init__ initializes it. When you override __new__, you usually call:

super().__new__(cls)
to actually create the object.

Here are some clear examples 👇

1. Basic __new__ override
class MyClass:
    def __new__(cls):
        print("Creating instance (in __new__)")
        instance = super().__new__(cls)
        return instance

    def __init__(self):
        print("Initializing instance (in __init__)")

obj = MyClass()
Output:
Creating instance (in __new__)
Initializing instance (in __init__)
👉 __new__ runs first, then __init__.

2. Passing arguments to __new__
class Person:
    def __new__(cls, name):
        print(f"__new__ called with {name}")
        instance = super().__new__(cls)
        return instance

    def __init__(self, name):
        print(f"__init__ called with {name}")
        self.name = name

p = Person("John")
3. Singleton pattern (common real use case)
Only one instance is ever created:

class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("Creating new instance")
            cls._instance = super().__new__(cls)
        else:
            print("Reusing existing instance")
        return cls._instance

a = Singleton()
b = Singleton()

print(a is b)  # True
4. Immutable types (like int, str) use __new__
If you subclass an immutable type, you must use __new__:

class PositiveInt(int):
    def __new__(cls, value):
        if value < 0:
            value = 0
        return super().__new__(cls, value)

x = PositiveInt(-5)
print(x)  # 0
👉 Here, int is immutable, so modification must happen in __new__, not __init__.

5. Returning a different type (advanced)
class Factory:
    def __new__(cls, value):
        if isinstance(value, int):
            return super().__new__(cls)
        else:
            return "Not an instance!"

obj1 = Factory(10)
obj2 = Factory("hello")

print(obj1)  # instance
print(obj2)  # "Not an instance!"
👉 __new__ can return anything, not necessarily an instance of the class.

Key Takeaways
__new__(cls, ...) → creates the object

__init__(self, ...) → initializes it

Always use:

super().__new__(cls)
unless you have a special reason not to

Needed especially for:

immutable types

singletons

metaprogramming

```python
class Fraction(object):
    def __init__(self, n, d):
        self.num = n
        self.denom = d

    def __float__(self):
        return self.num / self.denom

    def __mul__(self, other):
        top = self.num * other.num
        bottom = self.denom * other.denom
        return Fraction(top, bottom)

    
    def reduce(self):
        def gcd(n,d):
            while d != 0:
                (d,n) = (n%d, d)
            return n
        if self.denom == 0:
            return None
        elif self.denom == 1:
            return self.num
        else:
            greates_common_divisor = gcd(self.num, self.denom)
            top = int(self.num/greates_common_divisor)
            bottom = int(self.denom/greates_common_divisor)
            return Fraction(top, bottom)
        def __str__(self):
            return f"{self.num}/{self.denom}"
a = Fraction(4, 1)
b = Fraction(3, 5)
ar = a.reduce()
br = b.reduce()
print(ar, type(ar))
print(br, type(br))

# c = ar * br

# print(c)          # 12/10
# print(float(c))
# print(c.reduce())
```
