


why fib(34) fibonacci function results in 11405773 recursive calls and the same fib(34) results in 65 recursive calls

Great question — this is all about how recursion is implemented.

You’re seeing two very different numbers of calls for the same fib(34) because you’re comparing two different approaches:

🔴 1. Naive recursive Fibonacci
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)
What happens here?
Each call splits into two more calls

Those calls split again… and again…

This creates a huge recursion tree.

👉 For fib(34):

Total calls ≈ 11,405,773

Time complexity ≈ O(2ⁿ) (exponential)

Why so many calls?
Because the function recomputes the same values many times:

Example:

fib(5) calls fib(4) and fib(3)

fib(4) again calls fib(3) and fib(2)

So fib(3) is computed multiple times ❌

🟢 2. Optimized Fibonacci (with memoization)
def fib(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib(n-1, memo) + fib(n-2, memo)
    return memo[n]
What changes?
Each value is computed only once

Results are stored in memo

👉 For fib(34):

Total calls ≈ 65

Time complexity ≈ O(n)

💡 Why 65 calls specifically?
You compute each Fibonacci number from 0 → 34 once → 35 values

Plus some extra quick “lookups” (when memo is hit)

Total ≈ ~2n → about 65

⚖️ Big difference
Approach	Calls	Complexity
Naive recursion	11,405,773	O(2ⁿ) ❌
Memoization	65	O(n) ✅
🧠 Key insight
The naive version is slow because of repeated work.

Memoization removes repetition → turning exponential time into linear time.

Let’s go through your code step by step in a simple way.

🔹 The function
def in_list(L, e):
    print(e, L)
    if len(L) == 1:
        return L[0] == e
    else:
        return in_list(L[1:], e)
💡 What this function is trying to do
It tries to check if an element e exists in a list L using recursion (calling itself).

🔍 Step-by-step explanation
1. print(e, L)
Every time the function runs, it prints:

the value you're searching (e)

the current list (L)

This helps you see how the list gets smaller each time.

2. if len(L) == 1:
This is the base case (when recursion stops).

👉 If the list has only one element:

return L[0] == e
If that element equals e → returns True

Otherwise → returns False

3. else: return in_list(L[1:], e)
This is the recursive step.

👉 L[1:] means:

“take the list without the first element”

So the function keeps calling itself with a smaller list.

⚠️ IMPORTANT PROBLEM in your code
Your function never checks the first element unless it's the last one.

So it skips values!

🔎 Example 1
test = [2,3,8,1]
print(in_list(test,1))
Calls look like this:
1 [2,3,8,1]
1 [3,8,1]
1 [8,1]
1 [1]
👉 Finally:

L[0] == 1 → True
✅ Works here by chance (because 1 is last)

❌ Example 2 (your mistake)
test = [2,5,1,8]
print(in_list(test,1))
Calls:
1 [2,5,1,8]
1 [5,1,8]
1 [1,8]
1 [8]
👉 Final check:

8 == 1 → False
❌ It returns False even though 1 is in the list!

✅ FIXED VERSION
You need to check the first element at every step:

def in_list(L, e):
    if len(L) == 0:
        return False
    if L[0] == e:
        return True
    return in_list(L[1:], e)
🧠 How the fixed version works
If list is empty → False

If first element matches → True

Otherwise → check the rest of the list

✔️ Now both cases work:
print(in_list([2,3,8,1],1))  # True
print(in_list([2,5,1,8],1))  # True
🔥 Key takeaway
Your original code uses recursion correctly

But it misses checking elements

Always check the current element before recursion
