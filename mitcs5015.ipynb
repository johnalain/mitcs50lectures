```python
def mult_recur(a,b):
        if b == 1:
                return a
        else:
                return a + mult_recur(a,b-1)
print(mult_recur(5,4))
```

```python
def power_recur(n,p):
        if p == 0:
            return 1
        # elif p==1:
        #     return n
        else:
            return n*power_recur(n,p-1)
print(power_recur(2,3))
```

```python
# factorial
def fact(n):
        if n == 1:
                return 1
        else:
                return n*fact(n-1)
print(fact(4))
```
