```python
# a better and cleaner way to do above cell code using dicyionary
# nice to use one data structure no separate lists
#nice to index item of interest directly
# a python dictionary has entries that map a key value
# a list:
# index  0     1      2      3
# item   1.    2      3      4
# a dictionary:
# key1     key2     key3     key4
# val1     val2     val3     val4
def find_grades(grades, students):
    Lnew = []
    for elem in students:
        grade = grades[elem]
        Lnew.append(grade)
    return Lnew
d = {'rita': 'B', 'josephine': 'C', 'michel':'B', 'therese':'D'}
print(find_grades(d, ['rita', 'josephine']))
```




what means hashable object

A hashable object in Python is something that can be turned into a fixed value called a hash (using the built-in hash() function), and that value does not change during its lifetime.

🔑 Simple idea:
A hashable object is:

Immutable (it cannot change)

Has a consistent hash value

✅ Examples of hashable objects:
int → 5

float → 3.14

str → "hello"

tuple (only if all elements inside are hashable) → (1, 2, 3)

print(hash(5))
print(hash("hello"))
print(hash((1, 2, 3)))
❌ Not hashable (unhashable):
list → [1, 2, 3]

dict → {"a": 1}

set → {1, 2, 3}

hash([1, 2, 3])  # ❌ ERROR
👉 Why? Because these objects are mutable (they can change), so their hash would become unreliable.

📌 Why hashable matters:
Hashable objects are required for:

Dictionary keys

Set elements

my_dict = {
    "name": "John",   # string → hashable ✅
    10: "number"      # int → hashable ✅
}

my_set = {1, 2, 3}  # elements must be hashable
🧠 Quick rule:
If it can change → ❌ not hashable
If it cannot change → ✅ probably hashable

```python

def find_frequent_word(word_dict):
        words = []
        highest = max(word_dict.values())
        for k, v in word_dict.items():
                if v == highest:
                        words.append(k)
        return (words, highest)
word_dict = {'rah': 2, 'ah': 3, 'rom': 1, 'mah': 2, 'ro': 1}
most_freq = find_frequent_word(word_dict)
print(most_freq)
```

```python
def occurs_often(word_dict, x):
        freq_list = []
        word_freq_tuple = find_frequent_word(word_dict):
        while word_freq_tuple[1]>x:
                word_freq_tuple = find_frequent_word(word_dict)
                freq_list.append(word_freq_tuple)
                for word in word_freq_tuple[0]:
                        del(word_dict[word])
        return freq_list
```
